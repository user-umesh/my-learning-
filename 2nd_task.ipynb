### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
# Write code here
num = int(input("enter the salary: "))
ctc = (num * 10)/100
da = (num * 5)/100
pf = (num * 3)/100
slry = num / 100000
if 0 < slry <= 5:
  tax = 0
elif 5< slry <= 10:
  tax = (num * 10)/100
elif 10 < slry <= 20:
  tax = (num * 20)/100
elif slry > 20:
  tax = (num * 30)/100

gross = num - ctc - da - pf - tax
print(round(gross/12))
````

###`Problem 11`: A robot moves in a plane starting from the original point (0,0). The robot can move toward UP, DOWN, LEFT and RIGHT with a given steps.
The trace of robot movement is shown as the following:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
> The numbers after the direction are steps.

> `!` means robot stop there.

**Please write a program to compute the distance from current position after a sequence of movement and original point.**

*If the distance is a float, then just print the nearest integer.*

Example:

`Input`:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
`Output`:
```
2
```

````python
# Write code here
pos = [0,0] # here is the starting position
while True : # while loop for continusly taking the position
  path = input("enter the robot path : ")
  if path == "!" or path.upper() == "END":
    break
  else:
    direction = path.split()[0]
    steps = int(path.split()[1])

    if direction.upper() == "UP":
      pos[1] = pos[1] + steps
    elif direction.upper() == "DOWN":
      pos[1] = pos[1] - steps
    elif direction.upper() == "LEFT":
      pos[0] = pos[0] - steps
    elif direction.upper() == "RIGHT":
      pos[0] = pos[0] + steps
    else:
      pass
new_direction = round((pos[1]**2 + pos[0]**2)**0.5)
print(f"your new position is {pos}")
print("your distance from the origin is {}".format(new_direction))
````

###`Problem 13`:Print all the Armstrong numbers in a given range.
Range will be provided by the user<br>
Armstrong number is a number that is equal to the sum of power of its digits. For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.

````python
# Write code here

lower = int(input("enter the starting")) # getting the staring point

upper = int(input("enter the ending")) # getting the ending point

lis = [] # for storing the armstrong number

for i in range(lower,upper+1): # loop through the range

  num = i    # copying the integr

  var = 0   #  sum the numbers

  digit = len(str(i))

  for j in range(digit +1):  # loop through all the digit in number

    var += (num % 10)** digit  #getting each digit power

    num = num // 10
  if var == i:
    lis.append(i)
  else:
    continue
print(lis)
````
