### Q9:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

````python
print("hello")
````

````python
def gcd(a,b):
  divider_1 = []
  divider_2 = []
  for i in range(1,a+1):
    if a % i == 0:
      divider_1.append(i)
  for j in range(1,b+1):
    if b % j == 0:
      divider_2.append(j)
  cd = set(divider_1).intersection(divider_2)
  return abs(a*b) // max(cd)
print(gcd(12,18))
````

````python
# Write your code here
n_a , d_a = map(int,input("enter the first fraction").split())
n_b , d_b = map(int,input("enter the second fraction").split())
temp_n = (n_a * d_b) + (n_b * d_a)
temp_d = gcd(d_a,d_b)
print(f"{temp_n}/{temp_d}")
````
